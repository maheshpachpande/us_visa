```python
from airflow import DAG
from airflow.operators.python import PythonOperator
from airflow.providers.mongo.hooks.mongo import MongoHook
import pandas as pd
from datetime import datetime

from airflow.providers.mongo.hooks.mongo import MongoHook
import pandas as pd
from src.usVisaClassifier.utils import read_yaml_file

cnf = read_yaml_file("../config.yaml")

def fetch_mongo_data():
    hook = MongoHook(mongo_conn_id='mongo_default')
    collection = hook.get_collection(cnf['database']['collection_name'], cnf['database']['name'])
    cursor = collection.find({})  # You can add filters here
    df = pd.DataFrame(list(cursor))
    df.to_csv('artifact/data_ingestion/file/feature_store/raw_data1.csv', index=False)
    print(f"Read {len(df)} records from MongoDB")

default_args = {
    'owner': 'mahesh',
    'start_date': datetime(2025, 7, 22),
    'retries': 1
}

with DAG(
    dag_id='read_data_from_mongodb',
    default_args=default_args,
    schedule='@daily',  # ✅ use 'schedule' instead of 'schedule_interval'
    catchup=False
) as dag:

    read_data = PythonOperator(
        task_id='fetch_data',
        python_callable=fetch_mongo_data
    )

    
```
